```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
plt.figure(figsize=(12, 8))
for i, col in enumerate(df.columns[1:]):
    plt.subplot(len(df.columns), 3, i + 1) # plt.subplot(nrows, ncols, index)
    sns.histplot(df[col], bins=30, kde=False)
    plt.title(f"Historgram of {col}")

plt.tight_layout()
plt.plot()
```

```python
plt.figure(figsize=(6, 6))
sns.boxplot(y=df['age'], data=df, color='lightgreen')
plt.title(f"Boxplot of age")

plt.tight_layout()
plt.plot()
```
